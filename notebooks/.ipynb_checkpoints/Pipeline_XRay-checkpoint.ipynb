## Pipeline State Chain
```
rss_sources → selected_feed → raw_feed → feed_entries → selected_entry

  → embedding_text → embedding_vector → qdrant_ids

  → reflection_result → judge_result → timeline
  → knowledge_graph → story_summary → saved_story
  → meili_document → redis_key → api_response
```

````python
# ── [SETUP 3/4] Display helpers ──────────────────────────────────────────────
# Pure formatting utilities — zero business logic.

def _mem_mb():
    return psutil.Process(os.getpid()).memory_info().rss / 1024 / 1024 if _PSUTIL else 0.0

class _Timer:
    def __init__(self, label=""):
        self.label = label; self.elapsed_ms = 0.0
    def __enter__(self):
        self._t0 = time.perf_counter(); self._m0 = _mem_mb(); return self
    def __exit__(self, *_):
        self.elapsed_ms = (time.perf_counter() - self._t0) * 1000
        dm = _mem_mb() - self._m0
        s  = f"⏱ {self.label}: {self.elapsed_ms:.1f} ms"
        if _PSUTIL: s += f"  |  Δmem {dm:+.1f} MB"
        print(s)

def timer(label=""): return _Timer(label)

def box(text, title="", width=90):
    bar = "─" * width
    print(f"┌{bar}┐")
    if title:
        print(f"│ {title}"); print(f"├{bar}┤")
    for ln in (text or "(empty)").splitlines():
        for chunk in textwrap.wrap(ln, width - 2) or [""]:
            print(f"│ {chunk}")
    print(f"└{bar}┘")

def show_gateway_response(resp, label="GatewayResponse"):
    print(f"\n── {label} ─────────────────────────────────────")
    print(f"  provider     : {resp.provider}")
    print(f"  model        : {resp.model}")
    print(f"  latency_ms   : {resp.latency_ms:.1f}")
    print(f"  cost_usd     : ${resp.cost_usd:.6f}")
    print(f"  input_tokens : {resp.input_tokens}")
    print(f"  output_tokens: {resp.output_tokens}")
    print(f"  total_tokens : {resp.total_tokens}")
    print(f"  parsed type  : {type(resp.parsed).__name__ if resp.parsed else None}")
    print(f"  error        : {resp.error}")
    box(resp.content[:2000] + ("..." if len(resp.content) > 2000 else ""), title="raw content")

def show_manifest(manifest, label="Prompt Manifest"):
    print(f"\n── {label} ({manifest.stage} @ {manifest.version}) ──")
    print(f"  model         : {manifest.model_config.model}")
    print(f"  temperature   : {manifest.model_config.temperature}")
    print(f"  max_tokens    : {manifest.model_config.max_tokens}")
    print(f"  timeout_sec   : {manifest.model_config.timeout_seconds}")
    print(f"  cacheable     : {manifest.is_cacheable()}")
    print(f"  lifecycle     : {manifest.lifecycle_state}")
    box(manifest.system, title="system prompt")
    box(manifest.template, title="user template")

def show_vector(v, label="vector"):
    if not v: print(f"{label}: (empty)"); return
    n = len(v)
    head = ", ".join(f"{x:.6f}" for x in v[:5])
    tail = ", ".join(f"{x:.6f}" for x in v[-3:])
    norm = sum(x * x for x in v) ** 0.5
    print(f"{label} (dim={n}): [{head} … {tail}]")
    print(f"  min={min(v):.6f}  max={max(v):.6f}  L2-norm={norm:.6f}")

print("✓ Helper functions defined")
````

## Section 5 — Normalization
**Goal:** Apply production headline normalization and discovery scoring.  
**Functions:** `IngestionService.normalize_headline()` · `calculate_discovery_score()`  
**Output:** `normalized_headline`, `discovery_score`, `score_breakdown`


````python
# ── [5.1] normalize_headline ─────────────────────────────────────────────────
# Function  : IngestionService.normalize_headline(title)
# File      : app/services/ingestion_service.py
# Transforms: NFKC unicode → lowercase → strip BREAKING/LIVE → strip source suffixes
# Output    : normalized_headline (str)
from app.services.ingestion_service import IngestionService

print(f"RAW: {article_title_raw!r}")

with timer("normalize_headline"):
    normalized_headline = IngestionService.normalize_headline(article_title_raw)

print(f"NRM: {normalized_headline!r}")
````

````python
# ── [5.2] calculate_discovery_score ─────────────────────────────────────────
# Function  : IngestionService.calculate_discovery_score(title, content, pub_date, source_name)
# Output    : discovery_score (float), score_breakdown (dict)
import email.utils
from datetime import timezone

pub_date_raw = selected_entry.get("published", selected_entry.get("updated", ""))
pub_date     = None
try:
    _ts      = email.utils.parsedate_to_datetime(pub_date_raw)
    pub_date = _ts.astimezone(timezone.utc).replace(tzinfo=None)
    print(f"pub_date : {pub_date}")
except Exception as _e:
    print(f"⚠ Could not parse pub_date: {pub_date_raw!r} — {_e}")

source_name_for_score = selected_source.name if selected_source else "Unknown"

with timer("calculate_discovery_score"):
    discovery_score, score_breakdown = IngestionService.calculate_discovery_score(
        title=article_title_raw, content=article_content,
        pub_date=pub_date, source_name=source_name_for_score)

print(f"\ndiscovery_score : {discovery_score:.4f}")
for k, v in score_breakdown.items():
    print(f"  {k:<25} {v}")

if discovery_score < 0:
    print(f"\n⚠ REJECTED by discovery filter: {score_breakdown.get('reason')}")
    print("  Continuing for demonstration purposes.")
````

**Goal:** Compute content fingerprints and check the URL Bloom filter + PostgreSQL.  
**Functions:** `compute_fingerprints()` · `URLBloomFilter.is_seen()`  
**Output:** `fingerprints`, `is_duplicate`


````python
# ── [6.1] Compute fingerprints ───────────────────────────────────────────────
# Function  : compute_fingerprints(url, normalized_title, normalized_body)
# File      : app/core/fingerprint.py
# Output    : fingerprints (dict: url_hash, content_hash)
from app.core.fingerprint import compute_fingerprints

normalized_body_for_fp = IngestionService.normalize_headline(article_content[:500])

with timer("compute_fingerprints"):
    fingerprints = compute_fingerprints(
        url=canonical_url,
        normalized_title=normalized_headline,
        normalized_body=normalized_body_for_fp)

print(f"url_hash     : {fingerprints['url_hash']}")
print(f"content_hash : {fingerprints['content_hash']}")
````

````python
# ── [6.2] Bloom filter check ─────────────────────────────────────────────────
# Function  : url_bloom_filter.is_seen(url)
# File      : app/core/bloom_filter.py
# Output    : is_duplicate (bool)
from app.services.ingestion_service import url_bloom_filter

with timer("bloom_filter.is_seen"):
    is_duplicate = await url_bloom_filter.is_seen(canonical_url)

print(f"is_duplicate : {is_duplicate}")
if is_duplicate:
    print("  ⚠ In production: ingestion would stop here. Continuing for demo.")
````

````python
# ── [6.3] DB duplicate check (url_hash exact match) ─────────────────────────
from app.models.models import Article

async with async_session_factory() as _s:
    _r = await _s.execute(
        select(Article).where(Article.url_hash == fingerprints["url_hash"]).limit(1))
    db_duplicate_article = _r.scalar_one_or_none()

if db_duplicate_article:
    print(f"DB duplicate: id={db_duplicate_article.id}")
    print(f"  title      : {db_duplicate_article.title}")
    print(f"  created_at : {db_duplicate_article.created_at}")
else:
    print("No DB duplicate — article is new.")
````

## Section 7 — Article Persistence (PostgreSQL)
**Goal:** Build an `Article` ORM object and persist it to PostgreSQL.  
**Function:** `session.add(article)` → `flush()` → `commit()`  
**Output:** `article`, `article_id`


````python
# ── [7.1] Build Article ORM object ───────────────────────────────────────────
import uuid as _uuid_mod
from sqlalchemy import event as _sa_event

article = Article(
    id            = _uuid_mod.uuid4(),
    url           = canonical_url,
    url_hash      = fingerprints["url_hash"],
    content_hash  = fingerprints["content_hash"],
    title         = article_title_raw[:500] if article_title_raw else None,
    description   = str(selected_entry.get("summary", ""))[:1000] or None,
    content       = article_content[:50000] or None,
    author        = article_author[:200] if article_author else None,
    image_url     = article_image[:500] if article_image else None,
    source_id     = selected_source.id if selected_source else None,
    published_at  = pub_date,
    created_at    = datetime.now(UTC).replace(tzinfo=None),
    embedding_status = "pending",
    event_status     = "pending",
    cluster_status   = "pending",
)

print(f"id         : {article.id}")
print(f"title      : {article.title}")
print(f"source_id  : {article.source_id}")
print(f"content len: {len(article.content or '')}")
````

````python
# ── [7.2] Persist Article (with SQL capture) ──────────────────────────────────
sql_statements = []

@_sa_event.listens_for(engine.sync_engine, "before_cursor_execute")
def _cap_sql(conn, cursor, stmt, params, ctx, em):
    sql_statements.append({"sql": stmt[:400], "params": str(params)[:200]})

skip_persist = bool(db_duplicate_article)

if skip_persist:
    article    = db_duplicate_article
    article_id = db_duplicate_article.id
    print(f"⚠ Using existing article: {article_id}")
else:
    try:
        with timer("INSERT Article"):
            async with async_session_factory() as _s:
                _s.add(article)
                await _s.flush()
                await _s.commit()
        article_id = article.id
        print(f"✓ Persisted: {article_id}")
    except Exception as _e:
        print(f"✗ Insert failed: {_e} — loading existing")
        async with async_session_factory() as _s2:
            _r2 = await _s2.execute(
                select(Article).where(Article.url_hash == fingerprints["url_hash"]).limit(1))
            article    = _r2.scalar_one()
            article_id = article.id
        print(f"  Loaded existing: {article_id}")

try:
    _sa_event.remove(engine.sync_engine, "before_cursor_execute", _cap_sql)
except Exception:
    pass
````

````python
# ── [7.3] Show captured SQL statements ───────────────────────────────────────
print(f"SQL statements captured: {len(sql_statements)}")
for i, stmt in enumerate(sql_statements[-5:]):
    box(stmt["sql"].strip(), title=f"SQL [{i}]")
````

````python
# ── [7.4] Reload Article from DB (verify) ────────────────────────────────────
async with async_session_factory() as _s:
    _r = await _s.execute(select(Article).where(Article.id == article_id))
    article_from_db = _r.scalar_one()

print(f"article_from_db.id               : {article_from_db.id}")
print(f"article_from_db.embedding_status : {article_from_db.embedding_status}")
print(f"article_from_db.event_status     : {article_from_db.event_status}")
print(f"article_from_db.cluster_status   : {article_from_db.cluster_status}")
print(f"article_from_db.created_at       : {article_from_db.created_at}")
````

## Section 8 — Embeddings
**Goal:** Generate a 768-dimensional embedding vector for the article.  
**Function:** `embedding_service.get_embeddings([text])`  
**Output:** `embedding_text`, `embedding_vector`


````python
# ── [8.1] Prepare embedding text ─────────────────────────────────────────────
from app.services.embedding_service import embedding_service

parts = [
    article_title_raw or "",
    str(selected_entry.get("summary", ""))[:500],
    article_content[:4000],
]
embedding_text = " ".join(p.strip() for p in parts if p.strip())

print(f"embedding_text length  : {len(embedding_text)} chars")
print(f"preview                : {embedding_text[:200]}...")
````

````python
# ── [8.2] Check embedding Redis cache ────────────────────────────────────────
_emb_model       = settings.EMBEDDING_MODEL
_ck_hash         = hashlib.sha256(f"{_emb_model}:{embedding_text}".encode()).hexdigest()
embedding_cache_key = f"embed:{_ck_hash}"

print(f"embedding model     : {_emb_model}")
print(f"embedding_cache_key : {embedding_cache_key}")
_cached_emb = await cache_service.get(embedding_cache_key)
print(f"cache hit           : {_cached_emb is not None}")
````

````python
# ── [8.3] Generate embeddings ────────────────────────────────────────────────
# Function  : embedding_service.get_embeddings([embedding_text])
# Output    : embedding_vector (list[float])

with timer("embedding_service.get_embeddings"):
    embeddings_list  = await embedding_service.get_embeddings([embedding_text])

embedding_vector = embeddings_list[0] if embeddings_list else []
show_vector(embedding_vector, label="embedding_vector")
````

````python
# ── [8.4] Embedding statistics ───────────────────────────────────────────────
dim  = len(embedding_vector)
mean = sum(embedding_vector) / dim if dim else 0
std  = math.sqrt(sum((x - mean) ** 2 for x in embedding_vector) / dim) if dim else 0
l2   = math.sqrt(sum(x * x for x in embedding_vector))

print(f"dimension  : {dim}")
print(f"mean       : {mean:.6f}")
print(f"std dev    : {std:.6f}")
print(f"L2 norm    : {l2:.6f}  (1.0 = unit normalized)")
print(f"First 10   : {embedding_vector[:10]}")
````

## Section 13 — Story Clustering: Qdrant Candidate Search
**Goal:** Find existing stories semantically similar to this article.  
**Function:** `vector_service.search_similar(vector, score_threshold=0.80)`  
**Output:** `qdrant_candidates`, `candidate_clusters`


````python
# ── [13.1] Similarity search (SIMILARITY_THRESHOLD = 0.80) ───────────────────
SIMILARITY_THRESHOLD = 0.80

qdrant_candidates = []
if qdrant_healthy and embedding_vector:
    with timer(f"search_similar(threshold={SIMILARITY_THRESHOLD})"):
        qdrant_candidates = await vector_service.search_similar(
            vector=embedding_vector, limit=20, score_threshold=SIMILARITY_THRESHOLD)
    print(f"qdrant_candidates: {len(qdrant_candidates)}")
    for i, hit in enumerate(qdrant_candidates):
        p = hit.payload or {}
        print(f"  [{i:>2}] score={hit.score:.4f}  \"{str(p.get('title', ''))[:60]}\"")
else:
    print("⚠ Skipped: Qdrant not healthy.")
````

````python
# ── [13.2] Candidate article IDs ─────────────────────────────────────────────
candidate_article_ids = [
    hit.payload.get("article_id")
    for hit in qdrant_candidates
    if hit.payload and hit.payload.get("article_id") != str(article_id)
]
print(f"candidate_article_ids: {len(candidate_article_ids)}")
for cid in candidate_article_ids:
    print(f"  {cid}")
````

````python
# ── [13.3] Load candidate Articles from PostgreSQL ────────────────────────────
import uuid as _uuid2
from sqlalchemy.orm import selectinload

candidate_articles = []
if candidate_article_ids:
    _uuids = []
    for cid in candidate_article_ids[:10]:
        try:
            _uuids.append(_uuid2.UUID(str(cid)))
        except Exception:
            pass
    if _uuids:
        async with async_session_factory() as _s:
            _r = await _s.execute(
                select(Article)
                .where(Article.id.in_(_uuids))
                .options(selectinload(Article.source)))
            candidate_articles = list(_r.scalars().all())

print(f"candidate_articles: {len(candidate_articles)}")
for ca in candidate_articles:
    src = ca.source.name if ca.source else "?"
    print(f"  [{src:<20}] {ca.title}")
````

````python
# ── [13.4] Similarity score matrix ───────────────────────────────────────────
candidate_clusters = {}
for hit in qdrant_candidates:
    p = hit.payload or {}
    candidate_clusters[p.get("article_id", "?")] = {
        "score": hit.score, "title": p.get("title", "")}

print(f"candidate_clusters: {len(candidate_clusters)}  (threshold={SIMILARITY_THRESHOLD})")
for cid, d in sorted(candidate_clusters.items(), key=lambda x: -x[1]["score"]):
    bar   = "█" * int(d["score"] * 20)
    label = "✓ MERGE" if d["score"] >= SIMILARITY_THRESHOLD else "✗ skip"
    print(f"  {label}  score={d['score']:.4f}  {bar:<20}  \"{d['title'][:50]}\"")
````

````python
# ── [18.5] Cost estimate ─────────────────────────────────────────────────────
from app.ai.gateway import PRICING_TABLE

cfg     = synthesis_manifest.model_config
pricing = PRICING_TABLE.get(cfg.model, {})
print(f"Model      : {cfg.model}")
print(f"Fallbacks  : {cfg.fallback_models}")
print(f"Temp       : {cfg.temperature}")
print(f"Max tokens : {cfg.max_tokens}")

if pricing and filled_synthesis_prompt:
    est_in  = len(filled_synthesis_prompt) // 4
    est_out = cfg.max_tokens
    est_c   = (est_in * pricing["input"] + est_out * pricing["output"]) / 1_000_000
    print(f"\nEst. cost  : ~${est_c:.5f} USD ({est_in:,} in + {est_out:,} out tokens)")
````

````python
# ── [18.6] Intercept gateway & call ai_service.summarize_story_from_kg() ─────
# Function  : ai_service.summarize_story_from_kg(kg, contradictions, timeline, source_comparisons)
# File      : app/services/ai_service.py
# Output    : story_summary (StorySummaryResponse)
from app.services.ai_service import ai_service

raw_synthesis_response = None
_orig_gs2 = ai_gateway.generate_stage

async def _cap_synth(stage, prompt_variables, schema=None, story_id="", article_id=""):
    global raw_synthesis_response
    resp = await _orig_gs2(stage, prompt_variables, schema, story_id, article_id)
    if stage == "summary_generation":
        raw_synthesis_response = resp
    return resp

ai_gateway.generate_stage = _cap_synth

story_summary = None
try:
    with timer("ai_service.summarize_story_from_kg"):
        story_summary = await ai_service.summarize_story_from_kg(
            kg=kg_dict, contradictions=[], timeline=timeline, source_comparisons=[])
    print(f"✓ story_summary type: {type(story_summary).__name__}")
except Exception as _e:
    print(f"✗ Synthesis failed: {_e}")
    traceback.print_exc()
finally:
    ai_gateway.generate_stage = _orig_gs2
````

## Section 19 — Story Persistence (PostgreSQL)
**Goal:** Create and persist a Story ORM object with all related records.  
**Output:** `saved_story`


````python
# ── [19.1] Build Story ORM object ────────────────────────────────────────────
from app.models.models import Story, StoryArticle, StoryEntity, StoryTimelineEvent, StoryMetric

_now = datetime.now(UTC).replace(tzinfo=None)

story = Story(
    id               = _uuid_mod.uuid4(),
    headline         = story_summary.headline if story_summary else article_title_raw[:200],
    one_line_summary = story_summary.one_line_summary if story_summary else None,
    short_summary    = story_summary.short_summary if story_summary else None,
    detailed_summary = story_summary.detailed_summary if story_summary else None,
    key_facts        = story_summary.key_facts if story_summary else [],
    knowledge_graph  = kg_dict,
    story_status     = "active",
    first_seen_at    = _now,
    updated_at       = _now,
    cluster_confidence = qdrant_candidates[0].score if qdrant_candidates else None,
)

print(f"id       : {story.id}")
print(f"headline : {story.headline}")
````

````python
# ── [19.2] Build StoryArticle, StoryEntity, StoryTimelineEvent, StoryMetric ──

story_article_link = StoryArticle(
    id=_uuid_mod.uuid4(), story_id=story.id, article_id=article_id, created_at=_now)

story_entities_orm = [
    StoryEntity(id=_uuid_mod.uuid4(), story_id=story.id,
                entity_type=e.get("type", "UNKNOWN"), entity_value=e.get("value", ""))
    for e in entities[:15]
]

story_timeline_events_orm = []
for t in timeline:
    _edt = None
    if t["event_time"]:
        try:
            from dateutil import parser as _dup
            _edt = _dup.parse(t["event_time"]).replace(tzinfo=None)
        except Exception:
            pass
    story_timeline_events_orm.append(StoryTimelineEvent(
        id=_uuid_mod.uuid4(), story_id=story.id,
        event_time=_edt, event_time_raw=t.get("event_time_raw"),
        description=t.get("description", "")[:2000]))

story_metric = StoryMetric(
    id=_uuid_mod.uuid4(), story_id=story.id, views=0, bookmarks=0, shares=0, clicks=0)

print(f"StoryArticle links    : 1")
print(f"StoryEntity records   : {len(story_entities_orm)}")
print(f"StoryTimeline records : {len(story_timeline_events_orm)}")
````

````python
# ── [19.3] Persist Story + all related records ───────────────────────────────
sql_statements.clear()

@_sa_event.listens_for(engine.sync_engine, "before_cursor_execute")
def _cap2(conn, cursor, stmt, params, ctx, em):
    sql_statements.append({"sql": stmt.strip()[:300], "params": str(params)[:100]})

saved_story = None
try:
    with timer("INSERT Story + related records"):
        async with async_session_factory() as _s:
            _s.add(story)
            _s.add(story_article_link)
            for se in story_entities_orm:
                _s.add(se)
            for ste in story_timeline_events_orm:
                _s.add(ste)
            _s.add(story_metric)
            await _s.flush()
            await _s.commit()
    saved_story = story
    print(f"✓ Story persisted: {story.id}")
except Exception as _e:
    print(f"✗ Story persist failed: {_e}")
    traceback.print_exc()
finally:
    try:
        _sa_event.remove(engine.sync_engine, "before_cursor_execute", _cap2)
    except Exception:
        pass

print(f"\nSQL statements executed: {len(sql_statements)}")
for i, s in enumerate(sql_statements[:6]):
    print(f"  [{i}] {s['sql'][:180]}")
````

## Section 21 — Redis Cache
**Goal:** Cache the story for fast API responses.  
**Function:** `cache_service.set(key, value, ttl=TTL_STORY)`  
**Output:** `redis_key`, `redis_cached`


````python
# ── [21.1] Build cache key & set in Redis ────────────────────────────────────
from app.services.cache_service import TTL_STORY

story_id_for_cache = str(saved_story.id) if saved_story else str(story.id)
redis_key          = cache_service.story_key(story_id_for_cache)

cache_payload = {
    "id"              : story_id_for_cache,
    "headline"        : story_summary.headline if story_summary else None,
    "one_line_summary": story_summary.one_line_summary if story_summary else None,
    "category"        : story_summary.category if story_summary else None,
    "_cached_at"      : _now.isoformat(),
}

redis_cached = False
if redis_healthy:
    with timer("cache_service.set"):
        redis_cached = await cache_service.set(redis_key, cache_payload, ttl=TTL_STORY)
    print(f"redis_key    : {redis_key}")
    print(f"TTL_STORY    : {TTL_STORY}s")
    print(f"redis_cached : {redis_cached}")
else:
    print("⚠ Redis not healthy.")
````

````python
# ── [21.2] Verify cache hit ──────────────────────────────────────────────────
cache_read_back = None
if redis_healthy:
    with timer("cache_service.get (verify)"):
        cache_read_back = await cache_service.get(redis_key)
    print(f"hit        : {cache_read_back is not None}")
    if cache_read_back:
        print(f"cached_at  : {cache_read_back.get('_cached_at')}")
        print(f"headline   : {cache_read_back.get('headline')}")
````

## Section 22 — API Serialization
**Goal:** Load the story with all relationships and serialize to `StoryDetailResponse`.  
**Function:** `StoryDetailResponse.model_validate(story_orm)`  
**Output:** `api_response` — what the frontend receives


````python
# ── [22.1] Load Story with all relationships from DB ─────────────────────────
from sqlalchemy.orm import selectinload as _sl

target_story_id = saved_story.id if saved_story else story.id

async with async_session_factory() as _s:
    _r = await _s.execute(
        select(Story)
        .where(Story.id == target_story_id)
        .options(
            _sl(Story.timeline_events),
            _sl(Story.entities),
            _sl(Story.tags),
            _sl(Story.source_coverage),
            _sl(Story.differences),
            _sl(Story.articles).selectinload(StoryArticle.article).selectinload(Article.source),
        ))
    story_for_api = _r.scalar_one_or_none()

print(f"story_for_api loaded: {story_for_api is not None}")
if story_for_api:
    print(f"  timeline_events : {len(story_for_api.timeline_events)}")
    print(f"  entities        : {len(story_for_api.entities)}")
    print(f"  articles        : {len(story_for_api.articles)}")
````

````python
# ── [22.2] Serialize to StoryDetailResponse ──────────────────────────────────
from app.schemas.story import StoryDetailResponse

api_response = None
if story_for_api:
    try:
        with timer("StoryDetailResponse.model_validate"):
            api_response = StoryDetailResponse.model_validate(story_for_api)
        print(f"✓ type     : {type(api_response).__name__}")
        print(f"  id       : {api_response.id}")
        print(f"  headline : {api_response.headline}")
    except Exception as _e:
        print(f"✗ Schema validation failed: {_e}")
        traceback.print_exc()
````

````python
# ── [22.3] Inspect api_response fields ───────────────────────────────────────
if api_response:
    print(f"headline          : {api_response.headline}")
    print(f"one_line_summary  : {api_response.one_line_summary}")
    print(f"category          : {api_response.category}")
    print(f"source_count      : {api_response.source_count}")
    print(f"timeline_events   : {len(api_response.timeline_events)}")
    print(f"entities          : {len(api_response.entities)}")
    print(f"articles          : {len(api_response.articles)}")
    print(f"tags              : {len(api_response.tags)}")
    print(f"first_seen_at     : {api_response.first_seen_at}")
    print(f"updated_at        : {api_response.updated_at}")
````
